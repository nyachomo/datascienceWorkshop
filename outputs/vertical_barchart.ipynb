```python
#IMPORT THE NECESSARY LIBRARY
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.patches import Patch
```

```python
df=pd.read_excel('datasets/nyeri.xlsx')
```

CLEANING THE DATASET

```python
df['student_name']=df['student_name'].astype('str')
df['student_adm_no']=df['student_adm_no'].astype('str')
df[['student_year','year']]=df[['student_year','year']].astype('Int64')

string_columns=df.select_dtypes(include=['object','string']).columns
df[string_columns]=df[string_columns].apply(
    lambda x: x.str.strip().str.upper()
)
```

```python
df['student_gender'].value_counts()
```

```python
df['student_gender']=df['student_gender'].replace({
    'F':'FEMALE',
    'M':'MALE',
    'FE4MALE':'FEMALE'
})
```

```python
gender_count=df['student_gender'].value_counts()
plt.bar(gender_count.index,gender_count.values,color=['#ff99ff','#aa80ff'])
plt.title('Student Distribution')
plt.suptitle('Student Distribution by gender',fontsize=10,fontfamily='Arial',
             fontweight='bold',color='blue',fontstyle='italic')
plt.xlabel('Gender',color='red')
plt.ylabel('No of students',color='violet')

for i, value in enumerate(gender_count.values):

    percentage = (value / gender_count.sum()) * 100

    plt.text(
        i,
        value + 20,
        f'{value} ({percentage:.0f}%)',
        ha='center',
        fontsize=10,
        fontweight='bold',
        color='green'
    )

# Create legend for each color
legend_elements = [
    Patch(facecolor='#ff99ff', label='MALE'),
    Patch(facecolor='#aa80ff', label='FEMALE')
]

plt.legend(
    handles=legend_elements,
    title='Gender',
    loc='upper right'
)

plt.figtext(0.3,0.98,'Fig:7',ha='left')
plt.show()
```
