```python
##COUNTING CLASSES AN TOTAL NUMBER OF STUDENT
student_counts=df.dropna(
    subset=['attachment_period_2',
            'attachment_start_year',
            'attachment_end_year']).groupby([
                                    'student_class',
                                    'student_year',
                                    'student_course',
                                    'attachment_period_2',
                                    'attachment_start_year',
                                    'attachment_end_year'
]).size().reset_index(name='student_count')
```

```python
duplicates_attachment_clases=student_counts[
    ~student_counts.duplicated(
    subset=['student_class',
            'student_course',
            'student_year'],keep=False)]
```

```python
#SAVE DUPLICATES CLASES
duplicates_attachment_clases.to_excel('datasets/duplicate_attachment_clases.xlsx',index=False)
```

```python
class_details = duplicates_attachment_clases[
    [
        'student_class',
        'student_course',
        'student_year',
        'attachment_start_year',
         'attachment_end_year',
        'attachment_period_2'
    ]
].copy()
```

```python
class_details = class_details.rename(
    columns={
        'student_year': 'class_student_year',
        'attachment_end_year': 'class_attachment_end_year',
        'attachment_start_year': 'class_attachment_start_year',
        'attachment_period_2': 'class_attachment_period_2'
    }
)
```

```python
df = df.merge(
    class_details,
    on=['student_class', 'student_course'],
    how='left'
)
```

```python
df['student_year'] = df['student_year'].fillna(
    df['class_student_year']
)
```

```python
df['attachment_start_year'] = df['attachment_start_year'].fillna(
    df['class_attachment_start_year']
)
```

```python
df['attachment_end_year'] = df['attachment_end_year'].fillna(
    df['class_attachment_end_year']
)
```

```python
df['attachment_period_2'] = df['attachment_period_2'].fillna(
    df['class_attachment_period_2']
)
```
