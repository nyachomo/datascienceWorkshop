```python
#IMPORT PANDAS AND OS
import pandas as pd
import os
```

```python
#GET CURRENT WORKING DIRECTORY
os.getcwd()
```

```python
#IMPORT THE DATASET
df=pd.read_excel('datasets/nyeri.xlsx')
df.head(3)
```

```python
#GETTING COLUMN NAMES
list(df.columns)
```

```python
filtered_students=df.loc[
    df['student_final_score'].notna(),
    ['student_name','student_class','student_course','student_gender','student_final_score']
]

filtered_students.head(20)
```

```python
#SORT BY ONE COLUMN
#SORT BY STUDENTS FINAL SCORE
filtered_students.sort_values(by='student_final_score')##SORT VALUES FROM ASCENDING 
```

```python
#SORT FROM DESCENDING
filtered_students.sort_values(
    by='student_final_score',
    ascending=False
)
```

```python
df.info()
```

```python
df['student_year']=df['student_year'].astype('Int64')
```

```python
df.info()
```

```python
df['year']=df['year'].astype('Int64')
```

```python
df.info()
```

```python
filtered_student_2=df.loc[
    df['student_final_score'].notna() & df['student_year'].notna() & df['year'].notna(),
    ['student_name','student_class','student_course','student_gender','student_year','year','student_final_score']
]

filtered_student_2
```

```python
filtered_student_2.sort_values(
    by='student_year',
    ascending=False
)
```

```python
filtered_student_2.sort_values(
    by='student_gender',
    ascending=False
)
```

```python
#SORT BY TWO OR MORE COLUMS
filtered_student_2.sort_values(
    by=['student_gender','student_year']
)
```

```python
##SORT BY TWO COLUMNS WITH DIFFERENT SORTING DIRECTIONS
filtered_student_2.sort_values(
    by=['student_gender','student_final_score'],
    ascending=[True,False]
)
```

```python
#FILTERING AND SORTING
df[
    df['placement_status']=='Placed'
].sort_values(
    by=['student_gender','student_final_score'],
    ascending=[False,True]
).head(5)
```

```python
df.loc[
    (df['placement_status']=='Placed') & df['student_final_score'].notna(),
    ['student_name','student_gender','student_final_score']
].sort_values(
    by=['student_final_score','student_gender'],
    ascending=[False,False]
)
```
