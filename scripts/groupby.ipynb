```python
import pandas as pd
import os
```

```python
os.getcwd()
```

```python
df=pd.read_excel('datasets/nyeri.xlsx')
```

```python
df.head(4)
```

```python
df.dtypes
```

```python
#CONVERTIND GRADING SCORE TO INT
df['student_workbehaviour_grading_score']=pd.to_numeric(df['student_workbehaviour_grading_score'],errors='coerce')
```

```python
df['student_gender'].value_counts()
```

```python
df['student_gender']=df['student_gender'].str.strip().str.upper()
```

```python
df['student_gender'].value_counts()
```

```python
df['student_gender']=df['student_gender'].map({
    'MALE':'MALE',
    'FEMALE':'FEMALE',
    'F':'FEMALE',
    'M':'MALE',
    'FE4MALE':'FEMALE'
})
```

```python
df['student_gender'].value_counts()
```

```python
df.groupby('student_gender').size()
```

```python
##GROUP BY GENDER AND FIND THE MEAN OF THE FINAL SCORE
df.groupby('student_gender')['student_final_score'].mean()
```

```python
df.groupby('student_gender')['student_final_score'].max()
```

```python
##MINIMUM VALUE PER GROUP
df.groupby('student_gender')['student_final_score'].min()
```

```python
##FINDING SUM OF THE TOTAL FINAL SCORE OF EACH GROUP
df.groupby('student_gender')['student_final_score'].sum()
```

```python
##GROUP BY AND COUNTS
df.groupby('student_gender').count()
```

```python
##GROUBY AND COMPUTE STATISTICAL SUMMARY
df.groupby('student_gender')['student_final_score'].agg([
    'count','min','mean','max','size','median','std','var','first','last','nunique'
])
```

```python
##FINDING THE VARIANCE ALONE
df.groupby('student_gender')['student_final_score'].last()
```

```python
df.groupby(['student_gender','student_year'])['student_final_score'].max()
```

```python
df.groupby(['student_gender','year','attachment_period'])['student_skills_grading_score'].agg([
    'count','size','max','min','median','std','var','first','last','nunique','mean'
])
```

```python
##FINDING THE PLACEMENT RATE
df.groupby('student_gender')['placement_status'].value_counts(normalize=True)*100
```

```python
df.groupby(['student_gender','year'])['placement_status'].value_counts(normalize=True).mul(100).round(2)
```

```python
##COUNT NUMBER OF STUDENT PER COMPNANY
df['company_name'].value_counts()
```

```python
df.groupby(['company_name','student_gender'])['student_final_score'].agg([
    'size','max','min','mean'
])
```

```python
##FINDING COMPANIES WITH THE MOST STUDENTS
df.groupby(['company_name','year']).size().sort_values(ascending=False)
```
