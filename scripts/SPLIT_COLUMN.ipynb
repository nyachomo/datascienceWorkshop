```python
#
df_subset=df.loc[
    df['placement_status']=='Placed',
    ['student_name','student_class','attachment_period','attachment_start_date']
]
df_subset
```

```python
#SPLTING ATTACHMENT PERIOD INTO ATTACHMENT_START_MONTH AND ATTACHMENT_END_MONTH
df_subset[['attachment_start_month','attachment_end_month']]=(df_subset['attachment_period'].str.split('-',expand=True))
```

```python
df_subset
```

```python
#SPLITING STUDENT CLASS INTO 3 : (institution,course_code,category)
df_subset[['institution','course_code',]]=(df_subset['student_class'].str.split('/',n=1,expand=True))
```

```python
df_subset
```

```python
#SPLITING STUDENTS NAME INTO FIRSTNAME AND OTHER NAMES
df_subset[['firstname','other_names']]=(df_subset['student_name'].str.split(' ',n=1,expand=True))
```

```python
df_subset
```

```python
#RSPLIT
df_subset[['firstname_2','other_names_2']]=(df_subset['student_name'].str.rsplit(' ',n=1,expand=True))
```
