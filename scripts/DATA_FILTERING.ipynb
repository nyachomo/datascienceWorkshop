```python
#IMPORT THE NYERI DATASETS
df=pd.read_excel('datasets/nyeri.xlsx')
```

```python
df.dtypes
```

```python
df['student_name']=df['student_name'].astype(str)
df['student_year']=pd.to_numeric(df['student_year'],errors='coerce').astype('Int64')
df['student_adm_no']=pd.to_numeric(df['student_adm_no'],errors='coerce').astype('Int64')
df['year']=pd.to_numeric(df['year'],errors='coerce').astype('Int64')
df['attachment_start_date']=pd.to_datetime(df['attachment_start_date'],errors='coerce')
df['attachment_end_date']=pd.to_datetime(df['attachment_end_date'],errors='coerce')
```

```python
df.dtypes
```

```python
df['student_name']=df['student_name'].str.strip().str.upper()
```

```python
string_cloumns=df.select_dtypes(include=['string']).columns
df[string_cloumns]=df[string_cloumns].apply(
    lambda col:col.str.strip().str.upper()
)
```
