```python
#IMPORT PANDAS AND OS
import pandas as pd
import os
```

```python
#GET CURRENT WORKING DIRECTORY
os.getcwd()
```

```python
#IMPORT THE DATASET
df=pd.read_excel('datasets/nyeri.xlsx')
df.head(3)
```

```python
#GETTING COLUMN NAMES
list(df.columns)
```

```python
filtered_students=df.loc[
    df['student_final_score'].notna(),
    ['student_name','student_class','student_course','student_gender','student_final_score']
]

filtered_students.head(20)
```

```python
#SORT BY ONE COLUMN
#SORT BY STUDENTS FINAL SCORE
filtered_students.sort_values(by='student_final_score')##SORT VALUES FROM ASCENDING 
```

```python
#SORT FROM DESCENDING
filtered_students.sort_values(
    by='student_final_score',
    ascending=False
)
```
