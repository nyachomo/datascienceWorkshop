1.0 Vertical Barchart

```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.patches import Patch
```

```python
#LOAD THE DATASET
df=pd.read_excel('datasets/nyeri.xlsx')
```

```python
#DATA CLEANING
string_columns=df.select_dtypes(include=['object','string']).columns
df[string_columns]=df[string_columns].apply(
    lambda x: x.str.strip().str.upper()
)

df['student_gender']=df['student_gender'].replace({
    'F':'FEMALE',
    'M':'MALE',
    'FE4MALE':'FEMALE'
})
```

```python
gender_count=df['student_gender'].value_counts()
```

```python
#DRAW THE CHART
plt.bar(
    gender_count.index,
    gender_count.values,
    color=['#e600e6','#6699ff']
    )

plt.title(
    'Distribution of students by gender',
    color='#3366cc',
    fontweight='bold',
    #fontstyle='italic',
    fontsize=10,
    fontfamily='verdana'
    )

plt.ylabel('No Of Student')
plt.xlabel('Student Category') 
plt.figtext(0.98,0.1,'Fig : 7',ha='right')  

colors=[
    Patch(facecolor='#e600e6',label='MALE'),
    Patch(facecolor='#6699ff', label='FEMALE')
]

plt.legend(
    handles=colors,
    title='GENDER',
    loc='upper right'
)

for i, value in enumerate(gender_count.values):
    percentage=(value/gender_count.sum())*100
    plt.text(
        i,
        value,
        f'{value} ({percentage:.2f}%)',
        color='#0039e6'
    )
```
